```python
import pandas as pd

from sklearn.cluster import KMeans

from sklearn.preprocessing import StandardScaler
```

```python
df= pd.read_csv("/content/Country-data.csv")
df
```

```python
print("First 5 records:")

print(df.head())
```

```python
X=df.drop('country', axis=1)
X
```

```python
scaler =StandardScaler()

X_scaled =scaler.fit_transform(X)

print(X_scaled[:5])
```

```python
# Initialize KMeans clustering (e.g., using 3 clusters)
kmeans = KMeans(n_clusters=3, random_state=42)

# Fit the model and assign cluster labels to the DataFrame
df['Cluster'] = kmeans.fit_predict(X_scaled)

print("\nNumber of countries in each cluster:")
print(df['Cluster'].value_counts().sort_index())
```

```python
new_sample = pd.DataFrame(

[[10, 40, 7, 35, 30000, 5, 75, 2, 15000]], columns=X.columns )

new_sample_scaled = pd.DataFrame(

scaler.transform(new_sample),
columns=X.columns

)

new_cluster = kmeans.predict(new_sample_scaled)
```

```python
X_scaled = pd.DataFrame(
    scaler.fit_transform(X),
    columns=X.columns
)

print(X_scaled.head())
```

```python
kmeans =KMeans(n_clusters=3,random_state=42, n_init=10)
```

```python
# Fit the initialized KMeans model to the scaled data
k_labels = kmeans.fit_predict(X_scaled)
df['Cluster'] = k_labels

print("\nCountries with Cluster:")
print(df[['country', 'Cluster']].head(15))
```

```python
print("\nNumber of countries in each cluster:")

print(df['Cluster'].value_counts().sort_index())
```
