```python
import pandas as pd

import numpy as np

import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler

from scipy.spatial.distance import pdist, squareform

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
```

```python
df =pd.read_csv("/content/fitness_health_tracking.csv")

print("First 5 records:")

print(df.head())
```

```python
print("\nDataset Shape:")
print (df.shape)

print("\nColumn Names: ")
print(df.columns)

print("\nMissing Values: ")
print(df.isnull().sum())
```

```python
features = [
'Daily_Steps',
'Calories_Burned',
'Workout_Duration',
'Hours_Sleep']
X = df[features]
print("\nSelected Features: ")
print (X.head())
```

```python
scaler =StandardScaler()
X_scaled = scaler.fit_transform(X)
print("\nScaled Data:")
print(X_scaled[:5])
```

```python
X_scaled_df = pd.DataFrame(
X_scaled,
columns=features
)
print("\nScaled Data:")
print (X_scaled_df.head())
```

```python
distance = pdist (
X_scaled_df,
metric='euclidean'
)

distance_matrix = squareform(distance)
print("\nDistance Matrix:")
print(distance_matrix[:5, :5])
```

```python
Z = linkage(
X_scaled_df,
method='single')

print("\nLinkage Matrix:")
print (Z[:10])
```

```python
linkage_df = pd.DataFrame(
Z,
columns=[
'Cluster 1',
'Cluster 2',
'Distance',
'Number of Points'
])
print("\nLinkage Matrix DataFrame:")
print(linkage_df.head (10))
```

```python
plt.figure(figsize=(12, 6))
dendrogram(
    Z,
   truncate_mode='lastp',
   p=20
)

plt.title("Hierarchical Clustering Dendrogram - Single Linkage")
plt.xlabel("Cluster")
plt.ylabel("Distance")
plt.show()
```

```python
df[ 'Cluster'] = fcluster (
    Z,
t=4,
criterion='maxclust')
```

```python
print("\nFitness Data with Cluster: ")
print(
    df
     [
       [
        'User_ID',
        'Daily_Steps',
        'Calories_Burned',
        'Workout_Duration',
        'Hours_Sleep',
        'Activity_Type',
        'Diet_Preference',
        'Gender',
        'Age_Group',
        'Cluster'
       ]
     ].head(20)
)
```

```python
plt.figure(figsize=(10, 6))
plt.scatter (
df['Daily_Steps'],
df['Calories_Burned'],
c=df['Cluster'])
plt.xlabel("Daily Steps")
plt.ylabel("Calories Burned")
plt.title("Fitness Data - Hierarchical Clustering using Single Linkage")
plt.show()
```
