```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
df = pd.read_csv("/content/TrafficDataset (2).csv")

print("Original Dataset:")

print(df.head())
```

```python
df['Hour'] = pd.to_datetime(df['Time']).dt.hour
df
```

```python
df[ 'Traffic Situation'] = df['Traffic Situation'].replace({
  'low' :'normal',
'Normal': 'Normal',
'High': 'Heavy',
'Heavy': 'Heavy'
})
df
```

```python
df.head(20)
```

```python
le=LabelEncoder()

df['Day of the week']=le.fit_transform(df ['Day of the week'])

df['Traffic Situation']=le.fit_transform(df['Traffic Situation'])

df.head(20)
```

```python
x=df.drop(columns=['Traffic Situation'])

y=df['Traffic Situation']
```

```python
x = df.drop(columns=['Traffic Situation', 'Time', 'Date'])
y = df['Traffic Situation']

selector=SelectKBest(score_func=f_classif, k=4)

x_selected= selector.fit_transform(x, y)

selected_features= x.columns[selector.get_support()]

print("\nSelected Features:")

print(selected_features)
```

```python
print('Selected Features:')
print(selected_features)
```

```python
X_selected = pd.DataFrame(
    x_selected,
    columns=selected_features
)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_selected,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion='gini',
    random_state=42
)
model.fit(X_train, y_train)
```

```python
root_index = model.tree_.feature[0]
root_node = X_selected.columns[root_index]
print("\nRoot node:")
print(root_node)
```

```python
y_pred = model.predict(X_test)
```

```python
accuracy=accuracy_score(y_test, y_pred)
print("\nAccuracy:",accuracy)
```

```python
print("\n confusion matrrix")
print(confusion_matrix(y_test,y_pred))
```

```python
new_traffic = X_selected.iloc[[0]]
prediction = model.predict(new_traffic)
if prediction[0]==1:
  print("new traffic prediction:normal")
else:
  print("new traffic prediction:heavy")
```

```python
plt.figure(figsize(20,10))
plot_tree(model,
          feature_names=X_selected.columns,
          class_names=[str(c) for c in model.classes_])
```
